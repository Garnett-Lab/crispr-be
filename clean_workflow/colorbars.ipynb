```python
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import numpy as np

# Create colormap
colors = ['lightskyblue', 'blue']
cmap = mcolors.LinearSegmentedColormap.from_list('lightskyblue_blue', colors, N=100)

# Create figure with just colorbar
fig, ax = plt.subplots(figsize=(5, 0.5))
norm = mcolors.Normalize(vmin=3.0, vmax=10.0)
cbar = plt.colorbar(plt.cm.ScalarMappable(norm=norm, cmap=cmap), 
                    cax=ax, orientation='horizontal')
cbar.set_label('Odds-ratio value')

# Keep only first and last ticks
cbar.set_ticks([3.0, 10.0])

plt.savefig('figures/colorbar_oddsratio.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import numpy as np

# Define boundaries and colors with grey at 0
boundaries = [0, 0.3, 0.33, 0.5, 0.66, 1.0]
colors = ['grey', 'khaki', 'wheat', 'orange', 'darkorange']

# Calculate centered tick positions
tick_positions = [(boundaries[i] + boundaries[i+1]) / 2 for i in range(len(boundaries)-1)]

# Create discrete colormap
cmap = mcolors.ListedColormap(colors)
norm = mcolors.BoundaryNorm(boundaries, cmap.N)

# Create horizontal colorbar
fig, ax = plt.subplots(figsize=(4, 0.5))
cbar = plt.colorbar(plt.cm.ScalarMappable(norm=norm, cmap=cmap), 
                    cax=ax, orientation='horizontal',
                    boundaries=boundaries, 
                    ticks=tick_positions)
cbar.set_ticklabels(['0', '33%', '50%', '66%', '100%'])
cbar.set_label('Cell line hit')

plt.savefig('figures/colorbar_nhits.png', dpi=200, bbox_inches='tight')
plt.show()
```
